### Model matematyczny 

double alfa = 0.1;
double beta = 0.9;
double t=0;
int U0 = 1023;
int R = 1000;
double C;
void setup() {
Serial.begin(9600);
pinMode(7, OUTPUT); //pin rozładowujący
pinMode(6, INPUT); //pin ładujący
pinMode(A0, INPUT);
}
void loop(){
digitalWrite(7, LOW);
pinMode(6, INPUT);
pinMode(7, OUTPUT);
digitalWrite(6, LOW);
while (analogRead(A0)>U0*alfa){
}

pinMode(7, INPUT);
pinMode(6, OUTPUT);
digitalWrite(6, HIGH);
while (analogRead(A0)<alfa*U0){
}
t=micros();
while (analogRead(A0)<beta*U0){
}
t=micros()-t;
C=t/(R*log((1-alfa)/(1-beta)));
Serial.println(C);
delay(10);
}
